```python
import os
import pandas as pd

data_path = "../data/"

data_metadata = pd.read_csv(os.path.join(data_path, "images.csv"))
data_metadata = data_metadata[['image', 'label']]
```

```python
labels = data_metadata['label'].value_counts().head(11)
labels = labels.drop(labels.index[8])
print(labels)
```

```python
filtered = data_metadata[data_metadata['label'].isin(labels.index)]
filtered
```

```python
filtered.to_csv(os.path.join(data_path, "filter_data.csv"))
```
